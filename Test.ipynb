```python
import can
from can.interfaces.pcan import pcan

# 列出可用通道
print(can.detect_available_configs(interfaces='pcan'))
bus=can.Bus(interface="pcan",channel="PCAN_USBBUS1",bitrate=500000)

msg=can.Message(
    arbitration_id=0x202,
    data=[0,1,0,0,0,0,0,0],
    is_extended_id=False
)
bus.send(msg)

ack = bus.recv(timeout=1)
if ack:
    print(hex(ack.arbitration_id), ack.data)
bus.shutdown()
```
